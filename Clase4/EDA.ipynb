```python
# Cargar dataset
titanic = pd.read_csv("titanic_clase4.csv")
titanic
```

```python
#Exploración inicial
titanic.info()
```

```python
# mMuestra variables numericas con descripciones estadisticas básicas
titanic.describe()
```

```python
# contar nulos  en total por columna
titanic.isnull().sum()
```

```python
# se crea una copia de la base para ajustar
titanic_nuevo = titanic.drop(columns=["cabina"])
titanic_nuevo = titanic_nuevo.drop_duplicates()
```

```python
#¿ Que hacer con edad?
mediana_edad = titanic_nuevo["edad"].median()
print(f"mediana_edad:{mediana_edad}")
```

```python
#rellenar los valores nulos con la mediana
titanic_nuevo ["edad"]= titanic_nuevo["edad"].fillna(mediana_edad)
print(f"Filas y columnas después de limpiar:{titanic_nuevo.shape}")
titanic_nuevo.info()
```

```python
titanic_nuevo["edad"]
```

```python
# crear grupos de edades para otro análisis
titanic_nuevo["edad_grupo"] = pd.cut (
    titanic_nuevo["edad"],
    bins=[0,25,40,100],
    labels=["joven","adulto"," mayor"])

titanic_nuevo[["edad", "edad_grupo"]].head(10)
```
