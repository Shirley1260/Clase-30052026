```python
## Listado de diccionarios
inventario = [
    {"nombre": "Laptop Lenovo",    "precio": 2350000, "stock": 14, "categoria": "electrónica"},
    {"nombre": "Mouse Logitech",   "precio":   85000, "stock": 40, "categoria": "periféricos"},
    {"nombre": "Teclado mecánico", "precio":  320000, "stock":  3, "categoria": "periféricos"},
    {"nombre": "Monitor Samsung",  "precio":  950000, "stock":  8, "categoria": "electrónica"},
    {"nombre": "Silla ergonómica", "precio":  780000, "stock":  0, "categoria": "mobiliario"},
]

# 1. Agrega a CADA producto el campo "valor_inventario" (precio × stock)
for producto in inventario:
    producto["valor_inventario"]=producto["precio"]*producto["stock"]

# 2. Imprime nombre y valor_inventario de cada producto
for producto in inventario:
    print(producto["nombre"], producto["valor_inventario"])

# 3. Calcula el valor total de todo el inventario
total= 0
for producto in inventario:
    total += producto["valor_inventario"] # + es una suma que es lo mismo que decir total= total+producto(xxx)

print("valor total_inventario", total)

# 4. ¿Cuántos productos pertenecen a "electrónica"?
contador = 0

for producto in inventario:
    if producto["categoria"] == "electrónica":
        contador += 1

print(f"Cantidad de productos en electrónica: {contador}")
# 5. Lista los nombres de los productos con stock == 0
for producto in inventario:
    if producto["stock"] == 0:
        print(producto["nombre"])

# 6. Crea un diccionario llamado `resumen` con:
#    {"total_productos": ..., "valor_total": ..., "agotados": [...]}

# Total de productos
total_productos = len(inventario)

# Valor total del inventario
valor_total = sum(producto["valor_inventario"] for producto in inventario)

# Productos agotados
agotados = [producto["nombre"] for producto in inventario if producto["stock"] == 0]

# Diccionario resumen
resumen = {
    "total_productos": total_productos,
    "valor_total": valor_total,
    "agotados": agotados
}

print(resumen)
```

```python
##Ejercicios 8 - Funciones

# Crea las siguientes funciones y prueba cada una:

# 1. calcular_impuesto(precio, tasa=0.19)
#    Retorna el valor del impuesto (precio * tasa)

def calcular_impuesto(precio, tasa=0.19):
    """Retorna el valor del impuesto sobre el precio."""
    return precio * tasa

# Prueba
print(calcular_impuesto(1000))  # 190.0


# 2. precio_total(precio, descuento=0, tasa=0.19)
#    Retorna precio - descuento_en_pesos + impuesto
#    Usa calcular_impuesto() dentro

def precio_total(precio, descuento=0, tasa=0.19):
    """Retorna precio total después de descuento y sumando impuesto."""
    precio_desc = precio - descuento
    impuesto = calcular_impuesto(precio_desc, tasa)
    return precio_desc + impuesto

# Prueba
print(precio_total(1000, descuento=100))  # 1026.0


# 3. buscar_producto(inventario, nombre)
#    Recorre el inventario y retorna el diccionario del producto encontrado
#    Si no existe, retorna None
def buscar_producto(inventario, nombre):
    """Busca un producto por nombre en el inventario."""
    for producto in inventario:
        if producto["nombre"] == nombre:
            return producto
    return None

# Prueba
producto = buscar_producto(inventario, "Monitor Samsung")
print(producto)

# 4. productos_por_categoria(inventario)
#    Retorna un diccionario donde cada clave es una categoría
#    y el valor es una lista de nombres de sus productos
#    {"electrónica": ["Laptop Lenovo", "Monitor Samsung"], ...}

def productos_por_categoria(inventario):
    """Retorna un diccionario con categorías como claves y listas de nombres como valores."""
    categorias = {}
    for producto in inventario:
        cat = producto["categoria"]
        if cat not in categorias:
            categorias[cat] = []
        categorias[cat].append(producto["nombre"])
    return categorias

# Prueba
print(productos_por_categoria(inventario))
# 5. agregar_producto(inventario, nombre, precio, stock, categoria)
#    Crea el diccionario del nuevo producto y lo agrega al inventario
#    Retorna el inventario actualizado

def agregar_producto(inventario, nombre, precio, stock, categoria):
    """Crea un nuevo producto y lo agrega al inventario."""
    nuevo_producto = {
        "nombre": nombre,
        "precio": precio,
        "stock": stock,
        "categoria": categoria
    }
    inventario.append(nuevo_producto)
    return inventario

# Prueba
inventario = agregar_producto(inventario, "Auriculares Bose", 500000, 10, "periféricos")
print(inventario[-1])
```
