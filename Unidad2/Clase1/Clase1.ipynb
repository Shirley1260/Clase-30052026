```python
# Imputacion 1 salario ----- mediana 
mediana = df['salario_cop'].median()
df["salario_cop"].fillna(df["salario_cop"].median(), inplace=True)
df["salario_cop"].median()
```

```python
# Imputacion  2 salario --- media

estres = df['nivel_estres'].mean()
print(f'Media del nivel de estrés: {estres}')
df['nivel_estres'] = df['nivel_estres'].fillna(estres)
```

```python
# Imputacion  3 satisfaccion laboral --- media

estres = df['satisfaccion_laboral'].mean()
print(f'Media de la satisfacción laboral: {estres}')
df['satisfaccion_laboral'] = df['satisfaccion_laboral'].fillna(estres)
```

```python

df['satisfaccion_laboral'] = df['satisfaccion_laboral'].fillna(
df['satisfaccion_laboral'].median())
print(df['satisfaccion_laboral'].median())
```

```python
#Imputación ciudad moda
ciudad_moda = df['ciudad'].mode()[0]
print(f'Moda de la ciudad: {ciudad_moda}')
df['ciudad'] = df['ciudad'].fillna(ciudad_moda)


```

```python
# IQR = Q3 − Q1

# Límite inferior = Q1 − (1.5 × IQR)

# Límite superior = Q3 + (1.5 × IQR)


```

$$

IQR = Q3 - Q1

$$

$$

Limite\;superior = Q1 - (1.5*IQR)

$$

$$

Limite\;inferior = Q3 + (1.5*IQR)

$$




```python
# Cuartiles 
Q1 = df['horas_semana'].quantile(0.25)
Q3 = df['horas_semana'].quantile(0.75)
IQR = Q3 - Q1

print(f'Q1: {Q1}')
print(f'Q3: {Q3}')
print(f'IQR: {IQR}')
```

```python
Límite_inferior = Q1 - (1.5 * IQR)

Límite_superior = Q3 + (1.5 * IQR)
print(f"Límite_inferior: {Límite_inferior:.2f} horas")

print(f"Límite_superior: {Límite_superior:.2f} horas")
```

```python
outliers_iqr = df[
(df['horas_semana'] < Límite_inferior) |
(df['horas_semana'] > Límite_superior)
]

print(f"Outliers detectados en horas_semana: {len(outliers_iqr)}")
print(outliers_iqr[['id_trabajador', 'cargo', 'nivel', 'horas_semana']])
```
